```python
import sys
import os
import hydra
from hydra import initialize, compose
from omegaconf import OmegaConf

# [핵심 1] 프로젝트 루트 경로를 sys.path에 추가 (그래야 src를 import 할 수 있음)
# 현재 위치(notebooks)의 부모 폴더(..)를 절대 경로로 변환해 추가
sys.path.append(os.path.abspath(".."))


# [핵심 2] 노트북에서는 @hydra.main 대신 initialize와 compose를 사용합니다.
# 이미 초기화된 경우를 대비해 초기화 (재실행 시 에러 방지)
hydra.core.global_hydra.GlobalHydra.instance().clear()

# config_path는 현재 노트북 파일 기준 상대 경로입니다. (../config)
with initialize(version_base=None, config_path="../config"):
    cfg = compose(config_name="config")

print(f"불러온 모델 설정: {cfg.model.main_solver.model.alias} (별칭: {cfg.model.main_solver.model.alias})")
print(OmegaConf.to_yaml(cfg)) # 전체 설정 보고 싶으면 주석 해제
```
