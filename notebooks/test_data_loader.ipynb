```python
from pprint import pprint

def show_samples(data, n=5):
    if len(data) == 0:
        return
    sample = data[:n]
    for i, row in enumerate(sample, 1):
        print(f"======Sample {i}======\n{row}\n\n")


show_samples(test_data, n=5)
show_samples(validate_data, n=5)
```
